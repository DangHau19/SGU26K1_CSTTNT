# BÀI 4. TÔ MÀU ĐỒ THỊ (TOMAU.*)

## Mô tả bài toán:
Cho n thành phố, hãy tô màu các thành phố này sao cho không có bất kỳ hai thành phố nào kề nhau được tô cùng một màu và sử dụng số màu ít nhất có thể.

## Dữ liệu:
**Nhập từ bàn phím:**
- Dòng đầu tiên chứa số thành phố **n** (n ≤ 1000), m cần thành phố kề nhau.
- **m dòng tiếp theo**, mỗi dòng chứa 2 số u, v là 2 thành phố u, v kề nhau.

## Dữ liệu ra:
**Xuất ra màn hình:**
- Dòng đầu tiên chứa số màu k cần tô
- k dòng tiếp theo, liệt kê các thành phố được tô cùng một màu

## Ví dụ:

**Dữ liệu:**
```
9
1 2
1 4
2 4
2 7 8 9
1 3 5 6
```

**Kết quả:**
```
2
2 4 7 8 9
1 3 5 6
```

**Đồ thị:**
```
    2
   /|\
  1 4 7
  |   |
  3   8
     /
    9
```

## Thuật toán: Tham lam (Greedy Graph Coloring)

### Ý tưởng:
1. Xét từng đỉnh theo thứ tự từ 1 đến n
2. Với mỗi đỉnh, tô màu **nhỏ nhất** (bắt đầu từ 1) chưa bị sử dụng bởi các đỉnh kề
3. Lặp lại cho đến khi tô hết tất cả đỉnh

### Ưu điểm:
- Đơn giản, nhanh O(V + E)
- Tìm được lời giải (không đảm bảo tối ưu)

### Nhược điểm:
- Không đảm bảo số màu tối ưu (maximum clique bound)
- Kết quả phụ thuộc vào thứ tự xét đỉnh

````python
def to_mau_do_thi(n, edges):
    """
    Thuật toán Tham lam tô màu đồ thị.
    
    Args:
        n: Số đỉnh
        edges: Danh sách các cạnh (u, v)
    
    Returns:
        color_groups: Dictionary {màu: [danh sách đỉnh có màu này]}
    """
    # Khởi tạo danh sách kề
    adj = {i: [] for i in range(1, n + 1)}
    for u, v in edges:
        adj[u].append(v)
        adj[v].append(u)
    
    colors = {}  # Dictionary lưu màu của từng đỉnh {đỉnh: màu}
    
    for u in range(1, n + 1):
        # Lấy danh sách các màu đã bị các đỉnh kề "chiếm dụng"
        used_colors = {colors[v] for v in adj[u] if v in colors}
        
        # Tìm màu (bắt đầu từ 1) nhỏ nhất chưa bị chiếm
        c = 1
        while c in used_colors:
            c += 1
        
        colors[u] = c
    
    # Gom nhóm các đỉnh lại theo màu để dễ in ra
    color_groups = {}
    for u, c in colors.items():
        if c not in color_groups:
            color_groups[c] = []
        color_groups[c].append(u)
    
    return color_groups
````

## Ví dụ 1: Đồ thị từ dữ liệu đề bài

````python
# Dữ liệu từ ví dụ đề bài
n = 9
edges = [
    (1, 2), (1, 4),
    (2, 4), (2, 7), (2, 8), (2, 9),
    (1, 3), (1, 5), (1, 6)
]

# Gọi thuật toán
color_groups = to_mau_do_thi(n, edges)

# In ra kết quả
k = max(color_groups.keys()) if color_groups else 0
print(k)

for c in range(1, k + 1):
    if c in color_groups:
        print(" ".join(map(str, sorted(color_groups[c]))))
````

## Ví dụ 2: Đồ thị tam giác (3 đỉnh, 3 cạnh)

````python
# Tam giác: 1-2-3-1 (cần 3 màu)
n = 3
edges = [(1, 2), (2, 3), (3, 1)]

color_groups = to_mau_do_thi(n, edges)
k = max(color_groups.keys()) if color_groups else 0

print(f"Cần {k} màu:")
for c in range(1, k + 1):
    if c in color_groups:
        print(f"Màu {c}: {sorted(color_groups[c])}")
````

## Ví dụ 3: Đồ thị hai phía (Bipartite) - cần 2 màu

````python
# Đồ thị hai phía: 1-4, 2-4, 2-5, 3-5
n = 5
edges = [(1, 4), (2, 4), (2, 5), (3, 5)]

color_groups = to_mau_do_thi(n, edges)
k = max(color_groups.keys()) if color_groups else 0

print(f"Cần {k} màu:")
for c in range(1, k + 1):
    if c in color_groups:
        print(f"Màu {c}: {sorted(color_groups[c])}")
````

## Chương trình chính (nhập từ bàn phím):

````python
# Nhập dữ liệu từ bàn phím
try:
    # Nhập N (số đỉnh) và M (số cạnh)
    n, m = map(int, input("Nhập N M (cách nhau bằng dấu cách): ").split())
    
    # Nhập M cạnh
    edges = []
    for i in range(m):
        u, v = map(int, input(f"Nhập cạnh {i+1} (u v): ").split())
        edges.append((u, v))
    
    # Gọi thuật toán
    color_groups = to_mau_do_thi(n, edges)
    
    # Số lượng màu cần dùng chính là màu lớn nhất đã được tô
    k = max(color_groups.keys()) if color_groups else 0
    
    # In ra dòng đầu: Số màu k
    print(k)
    
    # In ra k dòng tiếp theo: Các đỉnh có cùng màu (đã sắp xếp tăng dần)
    for c in range(1, k + 1):
        if c in color_groups:
            print(" ".join(map(str, sorted(color_groups[c]))))
except:
    print(0)
````

## Giải thích chi tiết:

### 1. Greediness (Tham lam):
- **Ý tưởng**: Luôn chọn màu nhỏ nhất khả dụng
- **Lưu ý**: Không đảm bảo tối ưu (có thể cần nhiều màu hơn tối thiểu)
- **Ưu điểm**: Nhanh, đơn giản

### 2. Danh sách kề (Adjacency List):
```python
adj = {1: [2, 4, 3, 5, 6],
       2: [1, 4, 7, 8, 9],
       ...
}
```
- Lưu các đỉnh kề với mỗi đỉnh
- Duyệt nhanh các đỉnh kề

### 3. Used colors (Màu đã dùng):
```python
used_colors = {colors[v] for v in adj[u] if v in colors}
```
- Lấy tất cả màu của các đỉnh kề
- Tìm màu nhỏ nhất không nằm trong tập này

### 4. Độ phức tạp:
- **Thời gian**: O(V + E) hay O(V × Δ) (Δ = bậc tối đa)
- **Không gian**: O(V)

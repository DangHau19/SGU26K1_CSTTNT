# BÀI 1. LIỆT KÊ TỔ HỢP (TOHOP.*)

## Mô tả bài toán:
Cho tập hợp A có N phần tử đôi một khác nhau. Mỗi phần tử là một ký tự từ A → Z.
Hãy liệt kê các **tổ hợp của tập có N phần tử lấy K phần tử**.

## Dữ liệu:
**Nhập từ bàn phím:**
- Dòng đầu tiên chứa số **N K** (N là số lượng phần tử của tập hợp A, K là số phần tử cần lấy)
- **N dòng tiếp theo**, mỗi dòng chứa 1 ký tự từ A → Z (các ký tự đôi một khác nhau)

## Kết quả:
**Xuất ra màn hình:**
- Dòng đầu ghi số nguyên **M** là số lượng tổ hợp
- **M dòng tiếp theo**, mỗi dòng một tổ hợp của A

```python
def sinh_to_hop(k, n, elements):
    # [HÀM ĐẶC BIỆT]: Thuật toán Quay lui (Backtracking) cốt lõi
    # Thay vì duyệt 2 lần (1 lần đếm, 1 lần in), ta lưu tất cả vào mảng results.
    results = []
    current_indices = [-1] * k

    def Try(i, start_idx):
        # Điều kiện dừng (Base case): Khi đã chọn đủ k phần tử
        if i == k:
            # Map các chỉ số (index) đã chọn sang giá trị thực tế của tập hợp A
            combo = [elements[idx] for idx in current_indices]
            results.append(combo)
            return
        
        # Nhánh đệ quy: Thử chọn các phần tử tiếp theo (từ start_idx đến n-1)
        for j in range(start_idx, n):
            current_indices[i] = j  # Ghi nhận quyết định chọn phần tử ở vị trí j
            Try(i + 1, j + 1)       # Đi sâu vào nhánh tiếp theo, chỉ được chọn từ j+1 trở đi

    Try(0, 0)
    return results
```

## Hàm sinh tổ hợp sử dụng Backtracking:

```python
# Test với ví dụ từ đề bài
print("Bàn phím (Input):")
print("3 2")
print("F")
print("B")
print("A")
print("\nMàn hình (Output):")

n = 3
k = 2
elements = ['F', 'B', 'A']

tap_hop_ket_qua = sinh_to_hop(k, n, elements)

# In ra số lượng tổ hợp
print(len(tap_hop_ket_qua))

# In ra các tổ hợp
for combo in tap_hop_ket_qua:
    print(" ".join(combo))
```

## Nhập dữ liệu từ người dùng

```python
# Nhập N và K từ bàn phím
n, k = map(int, input("Nhập N và K (cách nhau bằng dấu cách): ").split())

# Nhập N phần tử của tập hợp
elements = []
for i in range(n):
    element = input(f"Nhập phần tử {i+1}: ").strip()
    elements.append(element)

# Gọi hàm xử lý
tap_hop_ket_qua = sinh_to_hop(k, n, elements)

# In ra dòng đầu tiên: Số lượng tổ hợp (M)
print(f"\nSố lượng tổ hợp: {len(tap_hop_ket_qua)}")

# In ra các dòng tiếp theo: Từng tổ hợp
print("\nCác tổ hợp:")
for combo in tap_hop_ket_qua:
    print(" ".join(combo))
```
